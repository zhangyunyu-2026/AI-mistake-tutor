```python
import random

# ----------------------
# 1. 规则模板库（错题分析/变式题/学习建议/鼓励语）
# ----------------------
ANALYSIS_TEMPLATES = [
    """
【考察知识点】
{知识点}

【错误原因分析】
你在解题过程中对{知识点}的概念理解不够清晰，尤其是在{易错点}环节出现了失误，没有正确运用{方法}。

【正确解题思路】
正确的做法是：首先明确{知识点}的核心规则，然后按照步骤1：{步骤1}，步骤2：{步骤2}，步骤3：{步骤3}进行计算，这样就能避免错误。
    """,
    """
【考察知识点】
{知识点}

【错误原因分析】
你没有掌握{方法}的使用条件，错误地将其应用在了不适用的场景中，导致计算过程出错。

【正确解题思路】
解决这类问题的关键是先判断题目是否符合{方法}的使用前提，再按照标准流程：{步骤1}→{步骤2}→{步骤3}来操作，就能得到正确结果。
    """
]

ADVICE_TEMPLATES = [
    """
建议1：重新复习{知识点}的定义和适用条件，整理成一份知识点清单。
建议2：多做3-5道同类型的基础题，重点练习{方法}的应用步骤。
建议3：建立错题本，把这道题的错误点和正确解法记录下来，定期回顾。
    """,
    """
建议1：针对{易错点}进行专项练习，找老师或同学请教这类题的解题技巧。
建议2：做题时养成先写解题步骤再计算的习惯，避免跳步出错。
建议3：做完题后主动检查一遍，重点核对关键步骤和计算结果。
    """
]

VARIANT_TEMPLATES = [
    """
变式题：用简便方法计算 {a} × {b} + {a} × {c}。
    """,
    """
变式题：用乘法分配律计算 {m} × ({n} + {p})。
    """
]

ENCOURAGE_TEMPLATES = [
    "错题是进步的阶梯，这次发现了{知识点}的薄弱点，下次一定能做对！加油💪",
    "没关系，谁都有犯错的时候，把这个知识点吃透，下次遇到同类题就不怕啦！冲鸭🦆",
    "你已经很棒了！能发现自己的错误就是进步的开始，多练几道题就能掌握啦！✨"
]

# 表情包库
EMOJI_LIST = [
    """
    (◕‿◕)
    /       \\
   |  加油！ |
    \\       /
     -------
    """,
    """
    (｡◕‿◕｡)
     /\\_/\\
    ( o.o )
     > ^ <
    冲鸭！
    """,
    """
    (≧∇≦)ﾉ
    ━━━━━━━━━
    你超棒的！
    ━━━━━━━━━
    """,
    """
    (๑•̀ㅂ•́)و✧
     /\\/\\/\\/\\
    努力向前冲！
    """,
    """
    (｀・ω・´)
    ──────────
    错题助你进步！
    ──────────
    """
]

# ----------------------
# 2. 核心功能实现
# ----------------------
def analyze_mistake(question_text):
    if "简便方法" in question_text and "×" in question_text:
        knowledge_point = "乘法分配律"
        error_point = "公式应用"
        method = "乘法分配律 a×b + a×c = a×(b+c)"
        step1 = "提取公因数a"
        step2 = "计算括号内的和"
        step3 = "计算最终结果"
    else:
        knowledge_point = "基础数学运算"
        error_point = "计算步骤"
        method = "四则运算规则"
        step1 = "明确运算顺序"
        step2 = "分步计算"
        step3 = "核对结果"
    
    template = random.choice(ANALYSIS_TEMPLATES)
    return template.format(
        知识点=knowledge_point,
        易错点=error_point,
        方法=method,
        步骤1=step1,
        步骤2=step2,
        步骤3=step3
    ).strip()

def generate_variant(original_question):
    a = random.randint(10, 30)
    b = random.randint(10, 50)
    c = random.randint(10, 50)
    m = random.randint(10, 30)
    n = random.randint(10, 50)
    p = random.randint(10, 50)
    
    template = random.choice(VARIANT_TEMPLATES)
    return template.format(a=a, b=b, c=c, m=m, n=n, p=p).strip()

def give_study_advice(analysis_result):
    if "乘法分配律" in analysis_result:
        knowledge_point = "乘法分配律"
        method = "简便运算"
        error_point = "公式应用"
    else:
        knowledge_point = "基础运算"
        method = "四则运算"
        error_point = "计算步骤"
    
    template = random.choice(ADVICE_TEMPLATES)
    return template.format(
        知识点=knowledge_point,
        方法=method,
        易错点=error_point
    ).strip()

def generate_encouragement():
    knowledge_point = "乘法分配律" if random.random() > 0.5 else "数学运算"
    return random.choice(ENCOURAGE_TEMPLATES).format(知识点=knowledge_point)

def get_random_emoji():
    return random.choice(EMOJI_LIST)

# ----------------------
# 3. 运行演示
# ----------------------
if __name__ == "__main__":
    print("🧠 AI 错题精灵已启动\n")
    
    # 示例错题（可替换为你自己的错题）
    sample_wrong = """
题目：用简便方法计算 17 × 23 + 17 × 47。
学生的错误解法：17 × 23 + 17 × 47 = 17 × (23 + 47) = 1190。
    """
    
    print("📝 正在分析错题...")
    analysis = analyze_mistake(sample_wrong)
    print("\n🔍 【错题分析结果】")
    print(analysis)
    
    print("\n📚 【学习建议】")
    advice = give_study_advice(analysis)
    print(advice)
    
    print("\n✏️ 【变式练习题】")
    variant = generate_variant(sample_wrong)
    print(variant)
    
    print("\n🎉 【鼓励彩蛋】")
    encourage = generate_encouragement()
    print(encourage)

    # 表情包放在所有内容最后，随机展示
    print("\n🎨 【鼓励表情包】（ASCII版）")
    print(get_random_emoji())
```
